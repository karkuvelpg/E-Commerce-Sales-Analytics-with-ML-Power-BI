# Sales Forecasting

```python
import pandas as pd
from prophet import Prophet
```

```python
df =pd.read_csv('..\data\clean_sales.csv')
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
df
```

```python
sales = df.groupby('InvoiceDate')['TotalPrice'].sum().reset_index()
sales
```

```python
sales['Date'] = sales['InvoiceDate'].dt.date
sales.drop(columns= "InvoiceDate",inplace = True)
sales
```

```python
sales = sales.groupby('Date')['TotalPrice'].sum().reset_index()
```

```python
sales
```

```python
sales.columns = ['ds', 'y']
```

```python
model = Prophet()
model.fit(sales)
```

```python
future = model.make_future_dataframe(periods = 90)
forecast = model.predict(future)
```

```python
forecast
```

```python
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].to_csv('..\data\sales_forecast.csv', index = False)
print("Sales forecasting complete. Saved to ..\data\sales_forecast.csv")
```
